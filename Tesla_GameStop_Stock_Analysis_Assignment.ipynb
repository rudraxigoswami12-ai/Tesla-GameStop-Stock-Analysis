# Question 1: Use yfinance to Extract Tesla Stock Data

Extract Tesla historical stock data, reset the index, save the dataframe, and display its first five rows.

```python
import yfinance as yf

# Extract Tesla stock data
tesla_data = yf.download("TSLA", period="max", auto_adjust=False)

# Reset the index
tesla_data.reset_index(inplace=True)

# Save the dataframe to a CSV file
tesla_data.to_csv("tesla_stock_data.csv", index=False)

# Display the first five rows
tesla_data.head()
```

# Question 2: Tesla Revenue Webscraping

This notebook extracts Tesla quarterly revenue from a public financial-statements webpage with `pandas.read_html()` and displays the last five rows of `tesla_revenue`.

```python
import pandas as pd

# Webscrape Tesla quarterly revenue data
url = "https://stockanalysis.com/stocks/tsla/financials/?p=quarterly"
headers = {"User-Agent": "Mozilla/5.0"}
tables = pd.read_html(url, storage_options=headers)

# Select the revenue row and display it as a dataframe
financials = tables[0]
revenue_row = financials[financials.iloc[:, 0].astype(str).str.contains("Revenue", case=False, na=False)]
if revenue_row.empty:
    raise ValueError("Revenue row was not found. The website layout may have changed.")
tesla_revenue = revenue_row.copy()

# Convert the row into a tidy dataframe with period and revenue columns
tesla_revenue = tesla_revenue.set_index(tesla_revenue.columns[0]).T.reset_index()
tesla_revenue.columns = ["Period", "Revenue"]
tesla_revenue["Revenue"] = tesla_revenue["Revenue"].astype(str)

# Display the last five rows
tesla_revenue.tail()
```

# Question 5: Tesla Stock Graph

Download Tesla stock data and use a `make_graph` function to plot its closing price with a clear graph title.

```python
import yfinance as yf
import pandas as pd
import plotly.graph_objects as go

# Download Tesla stock data and prepare the dataframe
tesla_data = yf.download("TSLA", period="5y", auto_adjust=False)
tesla_data = tesla_data.reset_index()
if isinstance(tesla_data.columns, pd.MultiIndex):
    tesla_data.columns = [col[0] if isinstance(col, tuple) else col for col in tesla_data.columns]
tesla_data["Date"] = pd.to_datetime(tesla_data["Date"])

# Define the make_graph function
def make_graph(stock_data, stock):
    fig = go.Figure()
    fig.add_trace(go.Scatter(
        x=stock_data["Date"], y=stock_data["Close"],
        mode="lines", name="Tesla Closing Stock Price"
    ))
    fig.update_layout(
        title="Tesla Stock Price Over the Last 5 Years",
        xaxis_title="Date",
        yaxis_title="Stock Price (USD)",
        template="plotly_white"
    )
    fig.show()

# Plot Tesla stock graph
make_graph(tesla_data, "Tesla")
```
